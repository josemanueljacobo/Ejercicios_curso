Ejercicio 2. Función calcular_promedio :

```python
def calcular_promedio (lista):
    if not lista:
        return 0
    return sum(lista)/ len(lista)

numeros = [10,20,30,40]
print("Promedio:", calcular_promedio(numeros))
 
```

Ejercicio 3. Función encontrar_duplicado

```python
def primer_duplicado (lista):
    elementos_vistos = set()
    for numero in lista:
      if numero in elementos_vistos:
         return numero
      return None
    
numeros = [3, 5, 1, 4, 3, 4, 8, 5]
print("Primer duplicado:", primer_duplicado(numeros))      
```

Ejercicio 4. Función enmascarado_datos :

```python
def enmascarar_variable(variable):
    cadena = str(variable)
    if len(cadena) <= 4:
        return cadena  # Si tiene 4 o menos caracteres, no enmascara
    return '#' * (len(cadena) - 4) + cadena[-4:]

dato = "123456789"
print(enmascarar_variable(dato))

  
```

Ejercicio 5. Función es_anagrama :

```python
def son_anagramas(palabra1, palabra2):
    return sorted(palabra1.lower()) == sorted(palabra2.lower())

# Ejemplo de uso
print(son_anagramas("amor", "roma"))
print(son_anagramas("Hola", "halo"))
print(son_anagramas("perro", "gato"))
```

Ejercicio 6. Función buscar_nombre :

```python
def buscar_nombre():
    # Solicitar al usuario ingresar los nombres separados por comas
    nombres = input("Ingresa una lista de nombres separados por coma: ").split(",")
    # Limpiar espacios adicionales alrededor de los nombres
    nombres = [nombre.strip() for nombre in nombres]

    # Solicitar un nombre para buscar
    nombre_buscar = input("Ingresa el nombre a buscar: ").strip()

    try:
        if nombre_buscar in nombres:
            print(f"El nombre '{nombre_buscar}' fue encontrado.")
        else:
            raise ValueError(f"El nombre '{nombre_buscar}' no se encuentra en la lista.")
    except ValueError as e:
        print(e)

```

Ejercicio 7. Función fibonacci :

```python
def fibonacci (n):
  if n < 0:
   raise ValueError("El indice no puede ser negativo")
  if n== 0:
    return 0
  elif n==1:
    return 1
  else:
    return fibonacci(n-1)+fibonacci(n-2)
n=3
print(f" El termino {n} de finonacci es: {fibonacci(n)}")  
  
```

Ejercicio 8. Función encontrar_puesto_empleado

```python
def buscar_puesto(nombre_completo, empleados):
    for empleado in empleados:
        nombre_empleado = f"{empleado['nombre']} {empleado['apellido']}"
        if nombre_empleado.lower() == nombre_completo.lower():
            return f"{nombre_completo} trabaja como {empleado['puesto']}."
    return f"{nombre_completo} no trabaja aquí."

# Caso de uso
empleados = [
    {'nombre': "Juan", 'apellido': "García", 'puesto': "Secretario"},
    {'nombre': "Mabel", 'apellido': "García", 'puesto': "Product Manager"},
    {'nombre': "Isabel", 'apellido': "Martín", 'puesto': "CEO"}
]

# Pruebas
print(buscar_puesto("Juan García", empleados)) 
print(buscar_puesto("Mabel García", empleados)) 
print(buscar_puesto("Carlos Pérez", empleados)) 
```

Ejercicio 14. No te vayas por las ramas :

```python
class Arbol:
    def __init__(self):
        self.tronco = 1  # Longitud inicial del tronco
        self.ramas = []  # Lista de ramas

    def crecer_tronco(self):
        self.tronco += 1
    
    def nueva_rama(self):
        self.ramas.append(1)  # Se añade una nueva rama con longitud 1
    
    def crecer_ramas(self):
        self.ramas = [rama + 1 for rama in self.ramas]  # Se incrementa la longitud de cada rama
    
    def quitar_rama(self, posicion):
        if 0 <= posicion < len(self.ramas):
            self.ramas.pop(posicion)  # Se elimina la rama en la posición indicada
        else:
            print("Posición no válida")
    
    def info_arbol(self):
        return {
            "longitud_tronco": self.tronco,
            "numero_ramas": len(self.ramas),
            "longitudes_ramas": self.ramas
        }

# Caso de uso
darbol = Arbol()
darbol.crecer_tronco()
darbol.nueva_rama()
darbol.crecer_ramas()
darbol.nueva_rama()
darbol.nueva_rama()
darbol.quitar_rama(2)
info = darbol.info_arbol()
print(info)
```

Ejercicio 15. Clase UsuarioBanco :

```python
class UsuarioBanco:
    def __init__(self, nombre, saldo, cuenta_corriente=True):
        self.nombre = nombre
        self.saldo = saldo
        self.cuenta_corriente = cuenta_corriente

    def retirar_dinero (self, cantidad):
        if cantidad > self.saldo:
            raise ValueError("Saldo insuficiente para realizar la operación.")
        
        self.saldo-= cantidad
    
    def transferir_dinero (self, otro_usuario, cantidad):
        if cantidad > self.saldo:
            raise ValueError("Saldo insuficiente para realizar la operación.")
        self.saldo-=cantidad
        otro_usuario.agregar_dinero (cantidad)
    
    def agregar_dinero (self, cantidad):
        self.saldo += cantidad

    def info_usuario (self):
        return {
            "nombre": self.nombre,
            "saldo" : self.saldo,
            "cuenta_corriente": self.cuenta_corriente
        }

# Caso de uso
Alicia = UsuarioBanco("Alicia", 100, True)
Bob = UsuarioBanco("Bob", 50, True)

Alicia.agregar_dinero(20)
Bob.transferir_dinero (Alicia,80)
Alicia.retirar_dinero (50)


print(Alicia.info_usuario())
print(Bob.info_usuario())


```
